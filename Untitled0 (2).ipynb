```python

from google.colab import files
from pypdf import PdfReader

uploaded = files.upload()

pdf_name = list(uploaded.keys())[0]

reader = PdfReader(pdf_name)

text = ""

for page in reader.pages:
    page_text = page.extract_text()
    if page_text:
        text += page_text + "\n"

print("PDF uploaded successfully!")
print("Total characters:", len(text))
print("\nDocument preview:\n")
print(text[:1000])
```

```python

def split_text(text, chunk_size=500, overlap=100):

    words = text.split()
    chunks = []

    start = 0

    while start < len(words):
        end = start + chunk_size

        chunk = " ".join(words[start:end])
        chunks.append(chunk)

        start += chunk_size - overlap

    return chunks


chunks = split_text(text)

print("Total chunks:", len(chunks))

for i, chunk in enumerate(chunks):
    print(f"\nChunk {i+1}:")
    print(chunk[:300])
```

```python

from sentence_transformers import SentenceTransformer
import numpy as np
import torch

device = "cuda" if torch.cuda.is_available() else "cpu"

print("Using device:", device)

embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2",
    device=device
)

embeddings = embedding_model.encode(
    chunks,
    convert_to_numpy=True,
    normalize_embeddings=True,
    show_progress_bar=True
)

print("\nEmbeddings created successfully!")
print("Embedding shape:", embeddings.shape)
```

```python

import faiss
import numpy as np

# Get embedding dimensions
dimension = embeddings.shape[1]

# Create FAISS index
vector_db = faiss.IndexFlatIP(dimension)

# Store embeddings
vector_db.add(
    np.asarray(embeddings, dtype="float32")
)

print("FAISS vector database created successfully!")
print("Total vectors stored:", vector_db.ntotal)
```

```python

from google import genai

api_key = "API KEY"

client = genai.Client(api_key=api_key)

print("Gemini connected successfully!")
```

```python

def ask_rag(question):

    # Step 1: Convert question into embedding
    question_embedding = embedding_model.encode(
        [question],
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    # Step 2: Search FAISS database
    scores, indices = vector_db.search(
        np.asarray(question_embedding, dtype="float32"),
        k=min(2, len(chunks))
    )

    # Step 3: Retrieve relevant chunks
    context = "\n\n".join(
        [chunks[i] for i in indices[0] if i != -1]
    )

    # Step 4: Prepare prompt for Gemini
    prompt = f"""
    You are a helpful AI document assistant.

    Answer the question using ONLY the context below.

    If the answer is not available in the context,
    say that the information is not available
    in the uploaded document.

    CONTEXT:
    {context}

    QUESTION:
    {question}

    ANSWER:
    """

    # Step 5: Generate answer
    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt
    )

    print("\nAI ANSWER:")
    print(response.text)


print("RAG AI Assistant is ready!")
```

```python
import gradio as gr

def smartbot(message, history):

    conversation = ""

    for chat in history:
        role = chat.get("role", "user")
        content = chat.get("content", "")
        conversation += f"{role}: {content}\n"

    prompt = f"""
    You are SmartBot, a friendly and helpful AI assistant.

    Previous conversation:
    {conversation}

    User's latest question:
    {message}

    Give a clear and helpful answer.
    """

    response = client.models.generate_content(
        model="gemini-3.1-flash-lite",
        contents=prompt
    )

    return response.text


demo = gr.ChatInterface(
    fn=smartbot,
    title="🤖 SmartBot - AI Chatbot",
    description="Welcome to SmartBot! Ask me anything."
)

print("Launching SmartBot...")
demo.launch(share=True)
```
